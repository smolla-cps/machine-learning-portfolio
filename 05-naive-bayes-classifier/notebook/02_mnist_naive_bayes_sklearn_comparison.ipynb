# MNIST Naive Bayes Model Comparison with scikit-learn

## Project Overview

This project compares multiple scikit-learn Naive Bayes formulations on the same normalized MNIST representation.

The supplied source material notes that scikit-learn provides several Naive Bayes implementations based on different distributional assumptions. The implementations evaluated here are:

- `GaussianNB`
- `MultinomialNB`
- `ComplementNB`
- `BernoulliNB`

Several Bernoulli configurations are also compared by changing the binarization threshold and smoothing parameter.

### Technical Coverage

- Gaussian Naive Bayes
- Multinomial Naive Bayes
- Complement Naive Bayes
- Bernoulli Naive Bayes
- Mean-based binarization
- Fixed 0.5 binarization
- Bernoulli smoothing parameter comparison
- Accuracy and confusion-matrix analysis

Reference: https://scikit-learn.org/stable/modules/naive_bayes.html

## 1. Environment

```python
import tensorflow as tf
import numpy as np
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.naive_bayes import GaussianNB, MultinomialNB, ComplementNB, BernoulliNB
```

## 2. MNIST Data Preparation

The same MNIST train/test split is normalized and flattened into 784-dimensional feature vectors so every estimator is evaluated on the same data representation.

```python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
x_train = np.float32(x_train/255).reshape(x_train.shape[0],-1)
x_test = np.float32(x_test/255).reshape(x_test.shape[0],-1)
```

## 3. Gaussian Naive Bayes

The first comparison uses `GaussianNB`.

```python
model =  GaussianNB()
model.fit(x_train, y_train)
pred = model.predict(x_test)
accuracy = accuracy_score(y_test, pred)
print(f'Accuracy = {accuracy:.4}')
cm = confusion_matrix(y_test, pred)
print(cm)
```

## 4. Multinomial Naive Bayes

The second model uses `MultinomialNB`.

```python
model =  MultinomialNB()
model.fit(x_train, y_train)
pred = model.predict(x_test)
accuracy = accuracy_score(y_test, pred)
print(f'Accuracy = {accuracy:.4}')
cm = confusion_matrix(y_test, pred)
print(cm)
```

## 5. Complement Naive Bayes

The third model uses `ComplementNB`.

```python
model =  ComplementNB()
model.fit(x_train, y_train)
pred = model.predict(x_test)
accuracy = accuracy_score(y_test, pred)
print(f'Accuracy = {accuracy:.4}')
cm = confusion_matrix(y_test, pred)
print(cm)
```

## 6. Bernoulli Naive Bayes

The default `BernoulliNB` configuration establishes the Bernoulli baseline.

```python
model =  BernoulliNB()
model.fit(x_train, y_train)
pred = model.predict(x_test)
accuracy = accuracy_score(y_test, pred)
print(f'Accuracy = {accuracy:.4}')
cm = confusion_matrix(y_test, pred)
print(cm)
```

## 7. Bernoulli Naive Bayes with Mean-Based Binarization

The binarization threshold is set to the mean value of the normalized training pixels.

```python
model =  BernoulliNB(binarize=np.mean(x_train))
model.fit(x_train, y_train)
pred = model.predict(x_test)
accuracy = accuracy_score(y_test, pred)
print(f'Accuracy = {accuracy:.4}')
cm = confusion_matrix(y_test, pred)
print(cm)
```

## 8. Bernoulli Naive Bayes with a 0.5 Threshold

A fixed binarization threshold of 0.5 provides a direct comparison with the mean-based threshold.

```python
model =  BernoulliNB(binarize=0.5)
model.fit(x_train, y_train)
pred = model.predict(x_test)
accuracy = accuracy_score(y_test, pred)
print(f'Accuracy = {accuracy:.4}')
cm = confusion_matrix(y_test, pred)
print(cm)
```

## 9. Bernoulli Naive Bayes with Adjusted Smoothing

The final configuration combines mean-based binarization with `alpha=0.05`.

```python
model =  BernoulliNB(alpha=0.05, binarize=np.mean(x_train))
model.fit(x_train, y_train)
pred = model.predict(x_test)
accuracy = accuracy_score(y_test, pred)
print(f'Accuracy = {accuracy:.4}')
cm = confusion_matrix(y_test, pred)
print(cm)
```

## 10. Comparative Results

| Model / Configuration | Saved Test Accuracy |
|---|---:|
| GaussianNB | 0.5558 |
| MultinomialNB | 0.8357 |
| ComplementNB | 0.7286 |
| BernoulliNB | 0.8413 |
| BernoulliNB + mean binarization | 0.8462 |
| BernoulliNB + 0.5 binarization | 0.8427 |
| BernoulliNB + mean binarization + alpha=0.05 | **0.8465** |

The saved results show that the Bernoulli configurations provide the strongest performance among the supplied scikit-learn comparisons, with the adjusted-smoothing configuration reaching 0.8465 test accuracy.

## 11. Key Findings

- Different Naive Bayes formulations can behave very differently on the same feature representation.
- The supplied MNIST comparison produces substantially lower accuracy with GaussianNB than with the Bernoulli and Multinomial formulations.
- Bernoulli performance changes with the binarization threshold.
- Adjusting the Bernoulli smoothing parameter produces the strongest saved scikit-learn result in the supplied comparison.
- Confusion matrices reveal the class-level structure behind the overall accuracy values.
